**Concept.** The *augmented matrix* [A | b] is the coefficient matrix A with the right-hand side b added as a last column.
Row operations on [A | b] do not change the solution set. The three operations are:
swap two rows, scale a row by c ≠ 0, and αR<sub>k</sub> + R<sub>j</sub> → R<sub>j</sub>. After row reduction, a *pivot* is the first
nonzero entry in a row. A matrix is in *reduced row echelon form* (RREF) when all four rules hold:

1. All-zero rows are at the bottom.
2. Every pivot is 1.
3. Every pivot is the only nonzero entry in its column.
4. Pivots move strictly to the right as you go down the rows.

## Topic 4: Solution Sets from RREF (Pivots, Free Variables, Consistency)

**Concept.** Row reduce the augmented matrix [A | b]. *Pivot columns* (the set **D**) hold the *dependent* variables.
The other columns (the set **F**) hold the *free* variables. A pivot in the last column means the row reads 0 = 1, so there
is **no solution**. A system is *consistent* if it has at least one solution. Let r be the number of nonzero rows of the
RREF (the same as the number of pivots). If the system is consistent, the number of free variables is n − r, where n is the
number of variables. As in lecture, F lists *all* non-pivot columns of [A | b]. For a consistent system it includes the last
column, which is not a variable.

SymPy's `.rref()` returns two things: the RREF matrix and a tuple of the pivot columns.

## Topic 5: Matrix Inverse by RREF

**Concept.** The *inverse* of a square matrix A is a matrix A⁻¹ with A A⁻¹ = A⁻¹ A = I, where I is the identity matrix.
To find it, write the augmented matrix [A | I] and row reduce. If the left block becomes I, the right block is A⁻¹:

[A | I] → [I | A⁻¹]

If the left block does **not** become I (for example, a row of zeros appears), A has **no inverse**. We call A *singular*.
The row operations that turn A into I are the same ones that turn I into A⁻¹.

**Careful:** in SymPy, `*` multiplies matrices (like `@` in NumPy). In NumPy, `*` works entry by entry, as you saw in Topic 1.

SymPy's `A.row_join(B)` places B to the right of A. The slice `R[:, 2:]` keeps all rows and the columns from index 2 on.

**Code to consider** (do not run it yet):

````python
A = sp.Matrix([[2, 1],
               [5, 3]])
M = A.row_join(sp.eye(2))      # the augmented matrix [A | I]
R, pivots = M.rref()
print(R)
print(pivots)
````

### Background: triangular matrices and inverses

A square matrix is **lower triangular** if every entry *above* the diagonal is 0. It is **upper triangular** if every entry *below*
the diagonal is 0. For example, [[2, 0, 0], [3, 1, 0], [4, 5, 6]] is lower triangular. Its diagonal entries are 2, 1, 6.

**Fact.** A lower triangular matrix L is invertible exactly when **every diagonal entry is nonzero**. When it is invertible, L⁻¹ is
lower triangular too. The same holds for upper triangular matrices. The exercise below tests this on two examples. The proof uses
determinants: det L is the product of the diagonal entries, and L is invertible exactly when det L ≠ 0.

**Why it matters.** Triangular matrices are easy to row reduce: with nonzero diagonal entries, each column gets a pivot on the diagonal.
They are also the factors in the LU factorization in the next notebook, so we need to know when they can be inverted.

**Worked example (2 × 2).** Row reduce [L | I] for L = [[2, 0], [3, 1]]:

- [L | I] = [[2, 0 | 1, 0], [3, 1 | 0, 1]]
- (1/2) R1 → R1 gives [[1, 0 | 1/2, 0], [3, 1 | 0, 1]]
- −3 R1 + R2 → R2 gives [[1, 0 | 1/2, 0], [0, 1 | −3/2, 1]]

The left block is I, so L⁻¹ = [[1/2, 0], [−3/2, 1]]. It is lower triangular.

**A matrix with a zero on the diagonal.** For M = [[1, 0], [1, 0]], the operation −1 R1 + R2 → R2 gives a row of zeros. The left block
of [M | I] cannot become I, so M has no inverse.
